```python
from pathlib import Path
import json
import pandas as pd

# Project paths
PROJECT_ROOT = Path.cwd().parent
RAW_DATA = PROJECT_ROOT / "data" / "raw"

print("Project root:", PROJECT_ROOT)
print("Raw data folder:", RAW_DATA)
```

```python
# Load the three CSV and JSON datasets
transactions = pd.read_csv(RAW_DATA / "track1_upi_transactions.csv")
kyc = pd.read_csv(RAW_DATA / "track1_kyc_records.csv")
merchants = pd.read_csv(RAW_DATA / "track1_merchants_master.csv")

with open(RAW_DATA / "track1_chargebacks.json", "r", encoding="utf-8") as f:
    chargebacks_data = json.load(f)

chargebacks = pd.DataFrame(chargebacks_data)

print("Transactions:", transactions.shape)
print("KYC:", kyc.shape)
print("Merchants:", merchants.shape)
print("Chargebacks:", chargebacks.shape)
```

```python
# Creating a dictionary to hold the datasets for easy access
datasets = {
    "transactions": transactions,
    "kyc": kyc,
    "merchants": merchants,
    "chargebacks": chargebacks,
}
```

```python
# Checking the structure of each dataset
for dataset_name, df in datasets.items():
    print("=" * 50)
    print(dataset_name.upper())
    print("=" * 50)
    print("Number of Rows:", len(df))
    print("Number of features:", len(df.columns))
    print("\nFeature Names:")
    print(df.columns.tolist())
    print("\nData type of each feature:")
    print(df.dtypes)
    print()
```

```python
# Checking for missing values in each dataset
for dataset_name, df in datasets.items():
    print("=" * 50)
    print(dataset_name.upper())
    print("=" * 50)

    missing = df.isna().sum()
    missing_percentage = (missing / len(df) * 100).round(2)

    profile = pd.DataFrame({
        "missing_count": missing,
        "missing_percentage": missing_percentage,
    })

    print(profile[profile["missing_count"] > 0])
    print()
```

```python
# Checking for completely duplicated rows/records in each dataset
for dataset_name, df in datasets.items():
    print(
        f"{dataset_name.title()}: "
        f"{df.duplicated().sum():,} number of completely duplicated rows/records."
    )
```

```python
# Displaying the first few rows/records of each dataset for respesentative view of the data
for dataset_name, df in datasets.items():
    print("=" * 50)
    print(dataset_name.upper())
    print("=" * 50)
    display(df.head(3))
```

```python
key_columns = {
    "transactions": ["txn_id", "user_id", "merchant_id", "utr"],
    "kyc": ["user_id"],
    "merchants": ["merchant_id"],
    "chargebacks": ["complaint_id", "txn_id", "user_id", "merchant_id"],
}

for dataset_name, columns in key_columns.items():
    df = datasets[dataset_name]

    print("=" * 70)
    print(dataset_name.upper())
    print("=" * 70)

    for column in columns:
        unique_count = df[column].nunique(dropna=True)
        duplicate_count = df[column].duplicated().sum()
        missing_count = df[column].isna().sum()

        print(
            f"{column:25} "
            f"unique={unique_count:,} | "
            f"duplicates={duplicate_count:,} | "
            f"missing={missing_count:,}"
        )
```

```python
for name, df, columns in [
    ("Transactions", transactions, ["txn_id", "user_id", "merchant_id", "utr"]),
    ("KYC", kyc, ["user_id"]),
    ("Merchants", merchants, ["merchant_id"]),
    ("Chargebacks", chargebacks, ["complaint_id", "txn_id", "user_id", "merchant_id"]),
]:
    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    for column in columns:
        print(f"\n--- {column} ---")
        print(df[column].dropna().astype(str).head(10).tolist())
```

```python
transaction_users = set(
    transactions["user_id"].dropna().astype(str)
)

kyc_users = set(
    kyc["user_id"].dropna().astype(str)
)

matched_users = transaction_users & kyc_users
unmatched_transaction_users = transaction_users - kyc_users

print("Unique transaction users:", len(transaction_users))
print("Unique KYC users:", len(kyc_users))
print("Users matching between Transactions and KYC:", len(matched_users))
print("Transaction users without KYC match:", len(unmatched_transaction_users))
```

```python
transaction_merchants = set(
    transactions["merchant_id"].dropna().astype(str)
)

merchant_master_ids = set(
    merchants["merchant_id"].dropna().astype(str)
)

matched_merchants = transaction_merchants & merchant_master_ids
unmatched_transaction_merchants = (
    transaction_merchants - merchant_master_ids
)

print("Unique transaction merchants:", len(transaction_merchants))
print("Unique merchant master IDs:", len(merchant_master_ids))
print(
    "Merchants matching Transactions → Master:",
    len(matched_merchants)
)
print(
    "Transaction merchants without master match:",
    len(unmatched_transaction_merchants)
)
```

```python
print("Chargebacks → Transactions")

chargeback_txns = set(
    chargebacks["txn_id"].dropna().astype(str)
)

transaction_ids = set(
    transactions["txn_id"].dropna().astype(str)
)

print("Unique chargeback transaction IDs:", len(chargeback_txns))
print("Matching transaction IDs:", len(chargeback_txns & transaction_ids))
print("Unmatched chargeback transaction IDs:", len(chargeback_txns - transaction_ids))


print("\nChargebacks → Users")

chargeback_users = set(
    chargebacks["user_id"].dropna().astype(str)
)

print("Matching user IDs:", len(chargeback_users & transaction_users))
print("Unmatched user IDs:", len(chargeback_users - transaction_users))


print("\nChargebacks → Merchants")

chargeback_merchants = set(
    chargebacks["merchant_id"].dropna().astype(str)
)

print(
    "Matching merchant IDs:",
    len(chargeback_merchants & transaction_merchants)
)

print(
    "Unmatched merchant IDs:",
    len(chargeback_merchants - transaction_merchants)
)
```
